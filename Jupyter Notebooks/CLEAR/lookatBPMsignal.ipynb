```python
pos = np.zeros([len(event_files),2,5])
charge = np.zeros([len(event_files),5])

for i in range(1,len(event_files)):
    f = h5py.File(event_files[i],'r')
    H_SA1 = f['CLEAREventData']['CA.BPM0530H-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,0,0] = np.sum(H_SA1[320:330])
    V_SA1 = f['CLEAREventData']['CA.BPM0530V-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,1,0] = np.sum(V_SA1[320:330])
    S_SA1 = f['CLEAREventData']['CA.BPM0530S-SA']['SamplesFromTrigger']['samples'][0]
    charge[i,0] = np.sum(S_SA1[320:330])
    
    H_SA2 = f['CLEAREventData']['CA.BPM0595H-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,0,1] = np.sum(H_SA2[320:330])
    V_SA2 = f['CLEAREventData']['CA.BPM0595V-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,1,1] = np.sum(V_SA2[320:330])
    S_SA2 = f['CLEAREventData']['CA.BPM0595S-SA']['SamplesFromTrigger']['samples'][0]
    charge[i,1] = np.sum(S_SA2[320:330])
    
    H_SA3 = f['CLEAREventData']['CA.BPM0690H-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,0,2] = np.sum(H_SA3[320:330])
    V_SA3 = f['CLEAREventData']['CA.BPM0690V-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,1,2] = np.sum(V_SA3[320:330])
    S_SA3 = f['CLEAREventData']['CA.BPM0690S-SA']['SamplesFromTrigger']['samples'][0]
    charge[i,2] = np.sum(S_SA3[320:330])
    
    H_SA4 = f['CLEAREventData']['CA.BPM0820H-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,0,3] = np.sum(H_SA4[320:330])
    V_SA4 = f['CLEAREventData']['CA.BPM0820V-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,1,3] = np.sum(V_SA4[320:330])
    S_SA4 = f['CLEAREventData']['CA.BPM0820S-SA']['SamplesFromTrigger']['samples'][0]
    charge[i,3] = np.sum(S_SA4[320:330])
    
    H_SA5 = f['CLEAREventData']['CA.BPM0890H-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,0,4] = np.sum(H_SA5[320:330])
    V_SA5 = f['CLEAREventData']['CA.BPM0890V-SA']['SamplesFromTrigger']['samples'][0]
    pos[i,1,4] = np.sum(V_SA5[320:330])
    S_SA5 = f['CLEAREventData']['CA.BPM0890S-SA']['SamplesFromTrigger']['samples'][0]
    charge[i,4] = np.sum(S_SA5[320:330])
```

```python
plt.plot(pos[1:,0,0], label='Hpos')
plt.plot(pos[1:,1,0], label='Vpos')
plt.plot(charge[1:,0], label='charge')
plt.legend()
plt.xlabel('Event Nr.')
plt.ylabel('Signal [mV]')
```
